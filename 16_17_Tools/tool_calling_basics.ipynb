```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

```python
# tool creation

@tool
def multiply(a: int, b: int) -> int:
  """Given two numbers a and b this tool return the multiplication of a and b"""
  return a * b
```

```python
print(multiply.invoke({'a': 10, 'b': 2}))
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
# tool binding

API_KEY = "AQ#########################"

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    google_api_key=API_KEY
)

llm_with_tool = llm.bind_tools([multiply])

llm_with_tool.invoke('Hi how are you?')
```

```python
# HumanMessage
query = HumanMessage('can you multiply 3 with 1000')
```

```python
messages = [query]
```

```python
# Tool Calling and getting AIMessage
result = llm_with_tool.invoke(messages)
result
```

```python
messages.append(result)
```

```python
# Tool execution and getting ToolMessage
tool_result = multiply.invoke(result.tool_calls[0])
tool_result
```

```python
messages.append(tool_result)
```
